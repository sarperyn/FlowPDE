## 1. Setup and Imports

```python
import sys
sys.path.append('..')

import torch
import matplotlib.pyplot as plt
import numpy as np
from torch.utils.data import DataLoader

# FlowPDE imports
from flowpde.datasets import PoissonForwardDataset
from flowpde.models import UNet
from flowpde.flows import FlowMatching
from flowpde.trainers import FlowMatchingTrainer

# Set random seed for reproducibility
torch.manual_seed(42)
np.random.seed(42)

# Device configuration
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"Using device: {device}")
```

## 2. Load Dataset

We use `PoissonForwardDataset` which returns:
- `u`: Solution field (target)
- `f`: Source term and coefficient (conditioning)

```python
# Configuration
data_path = '../data/poisson/forward/train.pt'
batch_size = 32

# Load dataset
dataset = PoissonForwardDataset(
    data_path=data_path,
    normalize=True,
    return_dict=True
)

# Create data loader
train_loader = DataLoader(
    dataset,
    batch_size=batch_size,
    shuffle=True,
    num_workers=0
)

print(f"Dataset size: {len(dataset)} samples")
print(f"Number of batches: {len(train_loader)}")

# Inspect a sample
sample = dataset[0]
print(f"\nSample data:")
print(f"  Solution (u) shape: {sample['u'].shape}")
print(f"  Condition (f) shape: {sample['f'].shape}")
```

## 3. Visualize Training Data

Let's visualize a few samples to understand the data

```python
# Get a batch of data
batch = next(iter(train_loader))
u_batch = batch['u']  # Solutions
f_batch = batch['f']  # Conditions (source + coefficient)

# Visualize first 4 samples
fig, axes = plt.subplots(4, 3, figsize=(12, 14))

for i in range(4):
    # Get grid size
    grid_size = int(np.sqrt(u_batch.shape[1]))
    
    # Reshape to 2D grids
    u = u_batch[i].reshape(grid_size, grid_size).numpy()
    f_data = f_batch[i].reshape(2, grid_size, grid_size).numpy()
    source = f_data[0]  # Source term
    coeff = f_data[1]   # Coefficient
    
    # Plot source term
    im1 = axes[i, 0].imshow(source, cmap='RdBu_r')
    axes[i, 0].set_title(f'Sample {i+1}: Source (f)')
    axes[i, 0].axis('off')
    plt.colorbar(im1, ax=axes[i, 0])
    
    # Plot coefficient
    im2 = axes[i, 1].imshow(coeff, cmap='viridis')
    axes[i, 1].set_title(f'Coefficient (κ)')
    axes[i, 1].axis('off')
    plt.colorbar(im2, ax=axes[i, 1])
    
    # Plot solution
    im3 = axes[i, 2].imshow(u, cmap='RdBu_r')
    axes[i, 2].set_title(f'Solution (u)')
    axes[i, 2].axis('off')
    plt.colorbar(im3, ax=axes[i, 2])

plt.tight_layout()
plt.show()

print("\nData statistics:")
print(f"  Solution range: [{u_batch.min():.3f}, {u_batch.max():.3f}]")
print(f"  Source range: [{f_batch[:, :grid_size**2].min():.3f}, {f_batch[:, :grid_size**2].max():.3f}]")
```

## 4. Create Flow Matching Model

We use a UNet architecture with time conditioning for the flow model

```python
# Model configuration
grid_size = int(np.sqrt(u_batch.shape[1]))
solution_dim = grid_size * grid_size
condition_dim = 2 * grid_size * grid_size  # source + coefficient

print(f"Grid size: {grid_size}x{grid_size}")
print(f"Solution dimension: {solution_dim}")
print(f"Condition dimension: {condition_dim}")

# Create UNet model
model = UNet(
    input_dim=solution_dim + condition_dim,  # Concatenated input
    base_channels=64,
    time_dim=32,
    num_blocks=4
)

# Count parameters
num_params = sum(p.numel() for p in model.parameters())
print(f"\nModel parameters: {num_params:,}")
```

## 5. Create Flow and Trainer

We use Flow Matching with linear interpolation paths

```python
# Create Flow Matching instance
flow = FlowMatching(
    model=model,
    path='linear',  # Can also use 'conditional_optimal_transport'
    sigma=0.0
)

# Create optimizer
learning_rate = 1e-4
optimizer = torch.optim.Adam(model.parameters(), lr=learning_rate)

# Create learning rate scheduler
epochs = 50
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=epochs)

# Create trainer
trainer = FlowMatchingTrainer(
    flow=flow,
    optimizer=optimizer,
    scheduler=scheduler,
    device=device
)

print("Flow Matching Trainer created successfully!")
print(f"Configuration: {trainer.get_flow_config()}")
```

## 6. Train the Model

Train using the flow matching algorithm

```python
# Training configuration
save_dir = '../results/training/poisson_forward_notebook'

print("Starting training...")
print("=" * 60)

# Train the model
trainer.train(
    data_loader=train_loader,
    epochs=epochs,
    print_interval=5,
    save_interval=10,
    save_dir=save_dir
)

print("\nTraining complete!")
```

## 7. Generate Samples

Use the trained flow to generate solutions conditioned on parameters

```python
# Set model to evaluation mode
model.eval()

# Get a test batch
test_batch = next(iter(train_loader))
true_solutions = test_batch['u'].to(device)
conditions = test_batch['f'].to(device)

# Generate samples using the flow
with torch.no_grad():
    # Sample from base distribution
    z = torch.randn_like(true_solutions).to(device)
    
    # Simple Euler integration for sampling
    num_steps = 50
    dt = 1.0 / num_steps
    x = z
    
    for i in range(num_steps):
        t = torch.ones(x.shape[0], 1, device=device) * (i * dt)
        # Predict velocity
        v = model(x, conditions, t)
        # Euler step
        x = x + v * dt
    
    generated_solutions = x

print(f"Generated {generated_solutions.shape[0]} samples")
print(f"True solution range: [{true_solutions.min():.3f}, {true_solutions.max():.3f}]")
print(f"Generated solution range: [{generated_solutions.min():.3f}, {generated_solutions.max():.3f}]")
```

## 8. Visualize Results

Compare true solutions with generated samples

```python
# Visualize comparison
num_samples = 4
fig, axes = plt.subplots(num_samples, 4, figsize=(16, 4*num_samples))

for i in range(num_samples):
    # Get data
    condition = conditions[i].cpu().reshape(2, grid_size, grid_size).numpy()
    source = condition[0]
    true_sol = true_solutions[i].cpu().reshape(grid_size, grid_size).numpy()
    gen_sol = generated_solutions[i].cpu().reshape(grid_size, grid_size).numpy()
    error = np.abs(true_sol - gen_sol)
    
    # Plot source
    im1 = axes[i, 0].imshow(source, cmap='RdBu_r')
    axes[i, 0].set_title(f'Sample {i+1}: Source (f)')
    axes[i, 0].axis('off')
    plt.colorbar(im1, ax=axes[i, 0])
    
    # Plot true solution
    im2 = axes[i, 1].imshow(true_sol, cmap='viridis')
    axes[i, 1].set_title('True Solution')
    axes[i, 1].axis('off')
    plt.colorbar(im2, ax=axes[i, 1])
    
    # Plot generated solution
    im3 = axes[i, 2].imshow(gen_sol, cmap='viridis')
    axes[i, 2].set_title('Generated Solution')
    axes[i, 2].axis('off')
    plt.colorbar(im3, ax=axes[i, 2])
    
    # Plot error
    im4 = axes[i, 3].imshow(error, cmap='Reds')
    axes[i, 3].set_title(f'Error (MAE: {error.mean():.4f})')
    axes[i, 3].axis('off')
    plt.colorbar(im4, ax=axes[i, 3])

plt.tight_layout()
plt.show()

# Compute metrics
mse = ((true_solutions - generated_solutions) ** 2).mean().item()
mae = (true_solutions - generated_solutions).abs().mean().item()

print(f"\nMetrics:")
print(f"  MSE: {mse:.6f}")
print(f"  MAE: {mae:.6f}")
```

## 9. Sample Multiple Solutions (Uncertainty Quantification)

Generate multiple samples for the same condition to visualize uncertainty

```python
# Select one condition
idx = 0
single_condition = conditions[idx:idx+1]
true_solution = true_solutions[idx:idx+1]

# Generate multiple samples
num_samples = 8
samples = []

with torch.no_grad():
    for _ in range(num_samples):
        # Sample different noise
        z = torch.randn_like(true_solution)
        x = z
        
        # Integrate
        for i in range(num_steps):
            t = torch.ones(1, 1, device=device) * (i * dt)
            v = model(x, single_condition, t)
            x = x + v * dt
        
        samples.append(x.cpu().numpy())

samples = np.array(samples).squeeze()

# Compute statistics
mean_solution = samples.mean(axis=0)
std_solution = samples.std(axis=0)

# Visualize
fig, axes = plt.subplots(2, 5, figsize=(20, 8))

# Plot source and true solution
source = single_condition[0].cpu().reshape(2, grid_size, grid_size).numpy()[0]
true = true_solution[0].cpu().reshape(grid_size, grid_size).numpy()

im = axes[0, 0].imshow(source, cmap='RdBu_r')
axes[0, 0].set_title('Source (f)', fontsize=12, fontweight='bold')
axes[0, 0].axis('off')
plt.colorbar(im, ax=axes[0, 0])

im = axes[1, 0].imshow(true, cmap='viridis')
axes[1, 0].set_title('True Solution', fontsize=12, fontweight='bold')
axes[1, 0].axis('off')
plt.colorbar(im, ax=axes[1, 0])

# Plot 4 random samples
for i in range(4):
    sample = samples[i].reshape(grid_size, grid_size)
    im = axes[0, i+1].imshow(sample, cmap='viridis')
    axes[0, i+1].set_title(f'Sample {i+1}', fontsize=12)
    axes[0, i+1].axis('off')
    plt.colorbar(im, ax=axes[0, i+1])

# Plot mean and std
mean_plot = mean_solution.reshape(grid_size, grid_size)
std_plot = std_solution.reshape(grid_size, grid_size)

im = axes[1, 1].imshow(mean_plot, cmap='viridis')
axes[1, 1].set_title('Mean Prediction', fontsize=12, fontweight='bold')
axes[1, 1].axis('off')
plt.colorbar(im, ax=axes[1, 1])

im = axes[1, 2].imshow(std_plot, cmap='Reds')
axes[1, 2].set_title('Std Deviation', fontsize=12, fontweight='bold')
axes[1, 2].axis('off')
plt.colorbar(im, ax=axes[1, 2])

# Plot error of mean
error = np.abs(true - mean_plot)
im = axes[1, 3].imshow(error, cmap='Reds')
axes[1, 3].set_title(f'Mean Error\n(MAE: {error.mean():.4f})', fontsize=12)
axes[1, 3].axis('off')
plt.colorbar(im, ax=axes[1, 3])

# Hide last subplot
axes[1, 4].axis('off')

plt.tight_layout()
plt.show()

print(f"\nUncertainty Quantification (from {num_samples} samples):")
print(f"  Mean prediction MAE: {np.abs(true - mean_plot).mean():.6f}")
print(f"  Average std: {std_plot.mean():.6f}")
print(f"  Max std: {std_plot.max():.6f}")
```

## 10. Summary

### What We Did

1. **Loaded Poisson forward problem data** - Parameters to solution mapping
2. **Created Flow Matching model** - UNet with time conditioning
3. **Trained using FlowMatchingTrainer** - Algorithm-centric trainer
4. **Generated samples** - Used ODE integration from noise to solution
5. **Quantified uncertainty** - Multiple samples from same condition

### Key Takeaways

- **Flow-based approach**: Learn probability distributions, not just point predictions
- **Uncertainty quantification**: Can generate multiple plausible solutions
- **Algorithm-centric**: Same trainer works for any PDE problem
- **Flexible**: Can switch to CNF or Rectified Flow with minimal changes
